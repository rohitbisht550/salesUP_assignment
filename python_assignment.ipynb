## standardized source column

```python
df["source"] = df["source"].str.strip().str.lower()
```

```python
df["source"] = df["source"].replace({
    "website": "Website",
    "web site": "Website",
    "linkedin": "LinkedIn",
    "cold call": "Cold Call",
    "referral": "Referral",
    "indeed": "Indeed"
})
```

## standardize status column

```python
df["status"] = df["status"].str.strip().str.lower()
df["status"] = df["status"].replace({
    "open":"Open",
    "won":"Won",
    "lost":"Lost"
})
```

## Fix date column "created_date"

```python
df["created_date"] = df["created_date"].str.strip()

df["created_date"] = pd.to_datetime(
    df["created_date"],
    format="mixed",
    dayfirst=False,
    errors="coerce"
)
```

## add month column

```python
df["month"] = df["created_date"].dt.month_name()
```

## handle values in deal_value column

```python
df["deal_value"] = (
    df["deal_value"]
    .astype(str)
    .str.replace(r"[₹,]", "", regex=True)
    .str.strip()
)
```

```python
df["deal_value"] = pd.to_numeric(df["deal_value"]).astype("Int64")
```

```python
df["deal_value"] = df["deal_value"].abs()
```

```python
df.loc[df["status"].isin(["Open", "Lost"]), "deal_value"] = df["deal_value"].fillna(0)
```

## Flag outliers

```python
df["value_outlier"] = df["deal_value"] > 200000
```

```python
df[df["value_outlier"]==True]
```

```python
won = df[(df["status"] == "Won") & (~df["value_outlier"])]
```

```python
won
```

## Monthly Summary

```python
df["is_won"] = df["status"] == "Won"
df["won_value"] = df["deal_value"].where(df["is_won"] & ~df["value_outlier"], 0).astype(float)
```

```python
monthly = df.groupby("month").agg(
    total_leads=("lead_id", "count"),
    won_leads=("is_won", "sum"),
    total_deal_value=("won_value", "sum")
)
monthly = monthly.sort_values(by = "total_deal_value", ascending = False)
monthly
```

```python
monthly["conversion_rate_%"] = (monthly["won_leads"] / monthly["total_leads"] * 100).round(1)
monthly = monthly.sort_values(by="conversion_rate_%", ascending=False)
monthly
```
